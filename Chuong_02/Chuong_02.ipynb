```python
from xml.dom.minidom import Document
doc = Document()

company = doc.createElement("company")
doc.appendChild(company)

def add_node(parent, tag, value):
    node = doc.createElement(tag)
    text = doc.createTextNode(value)
    node.appendChild(text)
    parent.appendChild(node)

add_node(company, "director", "Nguyen Van A")
add_node(company, "address", "123 Nguyen Trai, Ha Noi")
add_node(company, "phone", "0123456789")
add_node(company, "tax_code", "0101234567")

with open("company.xml", "w", encoding="utf-8") as f:
    f.write(doc.toprettyxml(indent="  "))

print("Đã tạo file company.xml")
```

BÀI 2.2

```python
import xml.etree.ElementTree as ET
students = ET.Element("students")

data = [
    {"id": "SV001", "name": "Nguyen Van A", "year": "2003", "class": "CNTT1", "gender": "Nam"},
    {"id": "SV002", "name": "Tran Thi B", "year": "2004", "class": "CNTT2", "gender": "Nữ"},
]

for sv in data:
    student = ET.SubElement(students, "student", id=sv["id"])
    ET.SubElement(student, "name").text = sv["name"]
    ET.SubElement(student, "year").text = sv["year"]
    ET.SubElement(student, "class").text = sv["class"]
    ET.SubElement(student, "gender").text = sv["gender"]

tree = ET.ElementTree(students)
tree.write("students.xml", encoding="utf-8", xml_declaration=True)

print("Đã tạo file students.xml")
```

```python
import requests
import xml.etree.ElementTree as ET
import csv

url = "http://www.hindustantimes.com/rss/topnews/rssfeed.xml"

response = requests.get(url)
response.encoding = "utf-8"         
xml_data = response.text

root = ET.fromstring(xml_data)

items = root.findall(".//item")

with open("rss_news.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["title", "link", "pubDate"])  # Header

    for item in items:
        title = item.findtext("title", default="")
        link = item.findtext("link", default="")
        pubDate = item.findtext("pubDate", default="")
        writer.writerow([title, link, pubDate])

print(" Đã tải RSS qua requests và lưu vào rss_news.csv!")
```

BÀI 2.10:

```python
import json

def thong_ke_nhan_vien(ten_tep):
    try:
        with open(ten_tep, 'r', encoding='utf-8') as f:
            data = json.load(f)

        ten_cong_ty = data.get("ten_cong_ty", "Không có")
        dia_chi = data.get("dia_chi", "Không có")
        danh_sach_don_vi = data.get("danh_sach_don_vi", [])

        tong_so_nhan_vien = 0
        for don_vi in danh_sach_don_vi:
            tong_so_nhan_vien += don_vi.get("so_nhan_vien", 0)

        print(f"Kết quả thực hiện chương trình:")
        print(f"Tên công ty: {ten_cong_ty}")
        print(f"Địa chỉ: {dia_chi}")
        print("----Thống kê nhân viên theo đơn vị----")

        if tong_so_nhan_vien == 0:
            print("Không có dữ liệu nhân viên để thống kê.")
            return

        for i, don_vi in enumerate(danh_sach_don_vi, start=1):
            ten_dv = don_vi.get("ten_don_vi", "N/A")
            so_nv = don_vi.get("so_nhan_vien", 0)
            
            ty_le_phan_tram = (so_nv / tong_so_nhan_vien) * 100
            
            print(f"{i}./Tên đơn vị: {ten_dv}")
            print(f"   - Số nhân viên: {so_nv}")
            print(f"   - Tỷ lệ: {ty_le_phan_tram:.2f}%") 

    except FileNotFoundError:
        print(f"Lỗi: Không tìm thấy tệp {ten_tep}")
    except json.JSONDecodeError:
        print(f"Lỗi: Tệp {ten_tep} không phải là tệp JSON hợp lệ.")
    except Exception as e:
        print(f"Đã xảy ra lỗi: {e}")

# Chạy chương trình
thong_ke_nhan_vien('cong_ty.json')
```
